```python
## Define PDF Processing Function

### Subtask:
Create a function `extract_text_from_pdf(pdf_file)` that takes a PDF file path as input, reads its content, and returns all text extracted from the PDF.
```

**Reasoning**:
The previous code block failed because it contained markdown text within a Python cell. To fix this, I will provide the actual Python code for the `extract_text_from_pdf` function, as specified in the subtask description. This function will use `PyPDF2` to read a PDF file and extract all its text.



```python
import PyPDF2

def extract_text_from_pdf(pdf_file):
    """
    Extracts text from a PDF file.

    Args:
        pdf_file (str): The path to the PDF file.

    Returns:
        str: All text extracted from the PDF.
    """
    text = ""
    try:
        with open(pdf_file, "rb") as file:
            reader = PyPDF2.PdfReader(file)
            for page_num in range(len(reader.pages)):
                page = reader.pages[page_num]
                text += page.extract_text() + "\n"
    except Exception as e:
        text = f"Error extracting text from PDF: {e}"
    return text

print("PDF text extraction function defined.")
```

## Define the Gradio Processing Function

### Subtask:
Create a function `process_input(file_input, text_input, action, question='')` that:
1. Takes an optional file input (PDF or text file), an optional text input, an action (`'summarize'` or `'qa'`), and an optional question for QA.
2. Extracts text from the file input if provided, otherwise uses the direct text input.
3. If the action is `'summarize'`, it uses the `summarizer` pipeline to summarize the extracted text.
4. If the action is `'qa'`, it uses the `question_answerer` pipeline with the extracted text and the provided question.
5. Returns the result (summary or answer).

**Reasoning**:
The previous markdown block outlined the subtask of creating the `process_input` function. Now, I will implement this function in Python, handling text extraction from files (PDF or text) or using direct text input, and then applying either the summarization or question-answering pipeline based on the specified action.



```python
import os

def process_input(file_input, text_input, action, question=''):
    """
    Processes input for summarization or question answering.

    Args:
        file_input (str): Path to an uploaded file (PDF or text).
        text_input (str): Direct text input.
        action (str): 'summarize' or 'qa'.
        question (str): Question for QA if action is 'qa'.

    Returns:
        str: The result (summary or answer) or an error message.
    """
    content = ""
    if file_input:
        if file_input.endswith('.pdf'):
            content = extract_text_from_pdf(file_input)
        elif file_input.endswith(('.txt', '.doc', '.docx')):
            try:
                with open(file_input, 'r', encoding='utf-8') as f:
                    content = f.read()
            except Exception as e:
                return f"Error reading text file: {e}"
        else:
            return "Unsupported file type. Please upload a PDF, TXT, DOC, or DOCX file."
    elif text_input:
        content = text_input
    else:
        return "Please provide either a file or text input."

    if not content or content.startswith("Error"): # Check if PDF extraction failed
        return content

    if action == 'summarize':
        if not content:
            return "No text provided for summarization."
        try:
            # Add truncation=True to handle long texts
            summary_result = summarizer(content, max_length=150, min_length=30, do_sample=False, truncation=True)
            return summary_result[0]['summary_text']
        except Exception as e:
            return f"Error during summarization: {e}"
    elif action == 'qa':
        if not question:
            return "Please provide a question for question answering."
        if not content:
            return "No context provided for question answering."
        try:
            # Add truncation=True to handle long contexts
            qa_result = question_answerer(question=question, context=content, truncation=True)
            return f"Answer: {qa_result['answer']} (Score: {qa_result['score']:.2f})"
        except Exception as e:
            return f"Error during question answering: {e}"
    else:
        return "Invalid action specified. Choose 'summarize' or 'qa'."

print("Gradio processing function `process_input` defined.")
```

**Reasoning**:
The previous step successfully imported the `gradio` library. Now, as per the subtask instructions, I need to define the Gradio interface using `gr.Interface`, including the file upload component and the tabbed structure for summarization and question answering.



```python
with gr.Blocks() as demo:
    gr.Markdown("# Document Analyzer")
    with gr.Row():
        file_input = gr.File(label="Upload Document (PDF, TXT, DOC, DOCX)", file_types=['.pdf', '.txt', '.doc', '.docx'])
        text_input = gr.Textbox(label="Or paste text here", lines=10)

    with gr.Tabs() as tabs:
        with gr.TabItem("Summarization", id="summarize_tab"):
            summarize_button = gr.Button("Summarize")
            summary_output = gr.Textbox(label="Summary", interactive=False)

        with gr.TabItem("Question Answering", id="qa_tab"):
            question_input = gr.Textbox(label="Your Question")
            qa_button = gr.Button("Get Answer")
            qa_output = gr.Textbox(label="Answer", interactive=False)

    # Connect components to the processing function
    summarize_button.click(fn=process_input, inputs=[file_input, text_input, gr.State('summarize')], outputs=summary_output)
    qa_button.click(fn=process_input, inputs=[file_input, text_input, gr.State('qa'), question_input], outputs=qa_output)

print("Gradio interface defined.")
```

**Reasoning**:
The Gradio interface has been defined in the previous step. To make it accessible and usable, I will now launch the Gradio application using `demo.launch()`.



```python
if demo.is_running: # Check if the app is already running
    demo.close() # Close the previous instance
demo.launch(debug=True)
print("Gradio application launched.")
```
